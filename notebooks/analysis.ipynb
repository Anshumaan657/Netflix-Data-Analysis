### Country Analysis

```python
plt.figure(figsize=(12,6))

sns.countplot(
    y='country',
    data=df,
    order=df['country'].value_counts().index[:10]
)

plt.title("Top 10 Countries Producing Netflix Content")

plt.show()
```

### Duration Analysis

```python
movies_df = df[df['type'] == 'Movie']
```

```python
movies_df['duration'].head()
```

```python
movies_df['duration'] = movies_df['duration'].str.replace(' min', '')
```

```python
plt.figure(figsize=(12,6))

sns.histplot(movies_df['duration'], bins=30)

plt.title("Distribution of Movie Durations")

plt.xlabel("Duration (minutes)")

plt.show()
```

### Netflix Growth Over Time

```python
df['date_added'].head(10)
```

```python
df['date_added'] = df['date_added'].str.strip()
```

```python
df['date_added'] = pd.to_datetime(df['date_added'])
```

```python
df['year_added'] = df['date_added'].dt.year
```

```python
plt.figure(figsize=(12,6))

sns.countplot(
    x='year_added',
    data=df,
    order=df['year_added'].value_counts().index
)

plt.xticks(rotation=45)

plt.title("Content Added to Netflix Each Year")

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(14,6))

sns.countplot(
    x='year_added',
    data=df,
    order=sorted(df['year_added'].dropna().unique())
)

plt.xticks(rotation=45)

plt.title("Content Added to Netflix Each Year")

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(14,6))

sns.countplot(
    x='year_added',
    hue='type',
    data=df
)

plt.xticks(rotation=45)

plt.title("Movies vs TV Shows Added Over the Years")

plt.tight_layout()

plt.show()
```

# Key Insights

1. Netflix experienced rapid content growth after 2015, indicating aggressive platform expansion.

2. Movies form the majority of Netflix’s content library, although TV Shows have also grown steadily in recent years.

3. Drama and International Movies are among the most dominant genres, reflecting Netflix’s focus on global audience engagement.

4. The United States contributes the highest amount of content on Netflix, making it the platform’s leading production region.

5. Most Netflix movies fall within the 90–120 minute duration range, suggesting a preference for standard feature-length content.

6. The increasing presence of international content highlights Netflix’s shift toward worldwide market expansion rather than region-specific entertainment.

7. Data visualization and exploratory analysis helped uncover meaningful trends hidden within raw streaming platform data.
